```python
class Account:
    def __init__(self, account_number, customer):
        self.account_number = account_number
        self.customer = customer
        self.balance = 0.0

    def deposit(self, amount):
        if amount <= 0:
            print("Error: Deposit amount must be positive.")
            return False
        self.balance += amount
        print(f"Successfully deposited ${amount:.2f} into Account {self.account_number}.")
        print(f"Current Balance: ${self.balance:.2f}")
        return True

    def withdraw(self, amount):
        if amount <= 0:
            print("Error: Withdrawal amount must be positive.")
            return False
        if self.balance - amount < 0:
            print(f"Error: Insufficient funds. Negative balance is not acceptable.")
            return False
        self.balance -= amount
        print(f"Successfully withdrew ${amount:.2f} from Account {self.account_number}.")
        print(f"Current Balance: ${self.balance:.2f}")
        return True


class Customer:
    def __init__(self, customer_id, name):
        self.customer_id = customer_id
        self.name = name
        self.accounts = []

    def add_account(self, account):
        self.accounts.append(account)


class BankSystem:
    def __init__(self):
        self.accounts = {}  # Maps account_number -> Account object
        self.customers = {}  # Maps customer_id -> Customer object

    def create_account(self):
        print("\n--- Create Account ---")
        try:
            account_num = int(input("Enter numeric Account Number: "))
            if account_num in self.accounts:
                print("Error: An account with this number already exists.")
                return

            cust_id = input("Enter Customer ID: ").strip()
            if not cust_id:
                print("Error: Customer ID cannot be empty.")
                return

            # Check if customer already exists, otherwise create a new one
            
            if cust_id in self.customers:
                customer = self.customers[cust_id]
                print(f"Existing customer found: {customer.name}")
            else:
                name = input("Enter Customer Name: ").strip()
                if not name:
                    print("Error: Customer Name cannot be empty.")
                    return
                customer = Customer(cust_id, name)
                self.customers[cust_id] = customer
                print(f"New customer profile created for {name}.")

            # Establish relations: Account belongs to one customer; Customer can have multiple accounts

            new_account = Account(account_num, customer)
            customer.add_account(new_account)
            self.accounts[account_num] = new_account
            print(f"Account {account_num} successfully created for {customer.name}.")

        except ValueError:
            print("Error: Account number must be a valid integer numeric value.")

    def deposit_account(self):
        print("\n--- Deposit Account ---")
        try:
            account_num = int(input("Enter Account Number: "))
            if account_num not in self.accounts:
                print("Error: Account not found.")
                return
            amount = float(input("Enter Deposit Value: "))
            self.accounts[account_num].deposit(amount)
        except ValueError:
            print("Error: Invalid numeric input.")

    def withdraw_account(self):
        print("\n--- Withdraw Account ---")
        try:
            account_num = int(input("Enter Account Number: "))
            if account_num not in self.accounts:
                print("Error: Account not found.")
                return
            amount = float(input("Enter Withdrawal Value: "))
            self.accounts[account_num].withdraw(amount)
        except ValueError:
            print("Error: Invalid numeric input.")

    def transfer_money(self):
        print("\n--- Transfer Money ---")
        try:
            from_account_num = int(input("Enter Source Account Number (Account1): "))
            if from_account_num not in self.accounts:
                print("Error: Source account not found.")
                return

            to_account_num = int(input("Enter Destination Account Number (Account2): "))
            if to_account_num not in self.accounts:
                print("Error: Destination account not found.")
                return

            if from_account_num == to_account_num:
                print("Error: Cannot transfer money to the same account.")
                return

            amount = float(input("Enter Transfer Value: "))
            if amount <= 0:
                print("Error: Transfer amount must be positive.")
                return

            source_account = self.accounts[from_account_num]
            dest_account = self.accounts[to_account_num]

            # Verify funds before performing the transaction

            if source_account.balance - amount < 0:
                print("Error: Insufficient funds in source account. Transfer cancelled.")
                return

            # Perform transfer
            
            source_account.balance -= amount
            dest_account.balance += amount

            print("\nTransfer Successful!")
            print(f"Account {from_account_num} Balance: ${source_account.balance:.2f}")
            print(f"Account {to_account_num} Balance: ${dest_account.balance:.2f}")

        except ValueError:
            print("Error: Invalid numeric input.")

    def account_balance(self):
        print("\n--- Account Balance ---")
        try:
            account_num = int(input("Enter Account Number: "))
            if account_num not in self.accounts:
                print("Error: Account not found.")
                return
            account = self.accounts[account_num]
            print(f"Account Holder: {account.customer.name} (ID: {account.customer.customer_id})")
            print(f"Account {account_num} Balance: ${account.balance:.2f}")
        except ValueError:
            print("Error: Account number must be a valid numeric integer.")

    def run(self):
        while True:
            print("\n=========================")
            print("      Menu Options:      ")
            print("=========================")
            print("1. Create Account: (Numeric)")
            print("2. Deposit Account: (Account, Value) + show the balance")
            print("3. Withdraw Account: (Account, Value) + show the balance")
            print("4. Transfer Money: (Account1, Account2, Value) + show the balances")
            print("5. Account balance: (Account) + show the balance")
            print("6. Quit")
            print("=========================")
            
            choice = input("Enter your choice (1-6): ").strip()
            
            if choice == '1':
                self.create_account()
            elif choice == '2':
                self.deposit_account()
            elif choice == '3':
                self.withdraw_account()
            elif choice == '4':
                self.transfer_money()
            elif choice == '5':
                self.account_balance()
            elif choice == '6':
                print("Exiting application. Thank you!")
                break
            else:
                print("Invalid option. Please choose a number between 1 and 6.")


bank_app = BankSystem()
bank_app.run()
```
